```python
def add(a, b):
  return a + b

def subtract(a, b):
  return a - b

def multiply(a, b):
  return a * b

def main():
  addition = add(10, 5)
  subtraction = subtract(10, 5)
  multiplication = multiply(10, 5)

  print("Addition:", addition)
  print("Subtraction:", subtraction)
  print("Multiplication:", multiplication)

main()
```
